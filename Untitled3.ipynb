```python
from sentence_transformers import CrossEncoder

reranker = CrossEncoder("cross-encoder/ms-marco-MiniLM-L-6-v2")
```

```python
def rerank(
    query: str,
    candidates: list[tuple[int, float, str]],
    top_n: int = 5
) -> list[tuple[int, float, str]]:
    if not 1 <= top_n <= len(candidates):
        raise ValueError("top_n must be between 1 and the number of candidates.")

    pairs = [(query, text) for _, _, text in candidates]
    scores = reranker.predict(pairs)

    ranked = sorted(
        zip(candidates, scores),
        key=lambda item: float(item[1]),
        reverse=True
    )

    return [
        (doc_id, float(score), text)
        for (doc_id, _, text), score in ranked[:top_n]
    ]
```

```python
class BM25:
    def __init__(self, docs: list[str], k1: float = 1.5, b: float = 0.75):
        if not docs:
            raise ValueError("Documents cannot be empty.")

        self.docs = docs
        self.k1 = k1
        self.b = b
        self.tokenized_docs = [tokenize(doc) for doc in docs]
        self.doc_lengths = [len(doc) for doc in self.tokenized_docs]
        self.N = len(docs)
        self.avgdl = sum(self.doc_lengths) / self.N

        self.term_freqs = []
        self.df = Counter()

        for tokens in self.tokenized_docs:
            counts = Counter(tokens)
            self.term_freqs.append(counts)
            self.df.update(counts.keys())

        self.idf = {
            term: math.log(1 + (self.N - df + 0.5) / (df + 0.5))
            for term, df in self.df.items()
        }

    def score_document(self, query_terms: list[str], doc_id: int) -> float:
        score = 0.0
        counts = self.term_freqs[doc_id]
        doc_length = self.doc_lengths[doc_id]

        for term in query_terms:
            tf = counts.get(term, 0)
            if tf == 0:
                continue

            denominator = tf + self.k1 * (
                1 - self.b + self.b * doc_length / self.avgdl
            )

            score += self.idf[term] * tf * (self.k1 + 1) / denominator

        return score

    def search(self, query: str, k: int = 3):
        if not 1 <= k <= self.N:
            raise ValueError("k must be between 1 and the number of documents.")

        query_terms = tokenize(query)
        scores = [
            self.score_document(query_terms, i)
            for i in range(self.N)
        ]

        ranked_ids = sorted(
            range(self.N),
            key=lambda i: (-scores[i], i)
        )

        return [
            (i, scores[i], self.docs[i])
            for i in ranked_ids[:k]
        ]
```

```python
def filter_detections(detection):
    filtered = []
    for item in detection:
        if item["confidence"] >= 0.8:
            filtered.append(item)
    return filtered
detections = [
    {"class": "helmet", "confidence": 0.95},
    {"class": "person", "confidence": 0.87},
    {"class": "helmet", "confidence": 0.72},
]
print(filter_detections(detections))
```

```python
from pathlib import Path

def read_docs(folder: str) -> dict[str,str]:
    folder_path = Path(folder)
    if not folder_path.is_dir():
        raise NotADirectoryError(f"The folder {folder_path} does not exist.")
    files = list(folder_path.glob("*.txt"))
    file_contents = {}
    for file_path in files:
        with open(file_path, "r", encoding="utf-8") as f:
            file_contents[file_path.name] = f.read()
    return file_contents
        
    

documents = read_docs("docs")

print(documents.keys())
print(documents["python.txt"])
```

```python
def chunk_text(text: str, size: int, overlap: int) -> list[str]:
    words = text.split()
    if size <= 0 or overlap < 0 or overlap >= size:
        return []
    return [
        " ".join(words[i:i + size])
        for i in range(0, len(words), size-overlap) 
    ]
text = "Python supports functions and classes for building applications"
print(chunk_text(text, 4, 1))
```

```python
from collections import Counter

def word_count_counter(text: str) -> list[tuple[str, int]]:
    counts = Counter(text.lower().split())
    return counts.most_common(10)
    
def word_count(text: str) -> list[tuple[str, int]]:
  counts = {}
  for word in text.lower().split():
    counts[word] = counts.get(word, 0) + 1

  sorted_words = sorted(counts.items(), key=lambda item: item[1], reverse=True)

  return sorted_words[:10]

text = """
Python makes development easier.
Python supports AI development.
AI uses data, and Python processes data.
"""

print(word_count(text))
print(word_count_counter(text))
```

```python
import numpy as np
embeddings = np.array([
    [3.0, 4.0],
    [5.0, 12.0],
    [8.0, 15.0]
])

norms = np.linalg.norm(embeddings, axis=1, keepdims=True)

normalized_embeddings = embeddings / norms

print(norms)
print(normalized_embeddings)
print(np.linalg.norm(normalized_embeddings, axis=1))
```

```python
import numpy as np
document = np.random.rand(100, 384)
query = np.random.rand(384)
score = document @ query
print(score.shape)
norms = np.linalg.norm(document, axis = 1, keepdims = True)
normalized = document / norms
normalized_norms = np.linalg.norm(normalized, axis=1)
is_normalized = np.allclose(normalized_norms, 1.0)
print(is_normalized)
top5 = np.argsort(score)[-5:][::-1]
print(top5)
print(score[top5])
```
